## 1. Load Final Evaluation Files

These files were generated by the model-development notebook. Here we only load and interpret them for the evaluation discussion.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

BASE_DIR = Path(r"C:\Users\13136\Desktop\Group13_Lab3_AAPL_EDA\Submission_By_Phase")
MODEL_RESULTS_DIR = BASE_DIR / "03_Model_Development_30pct" / "results"
PRESENTATION_DIR = BASE_DIR / "05_Presentation_Demonstration_20pct"

test_comparison = pd.read_csv(MODEL_RESULTS_DIR / "test_model_comparison.csv")
feature_importance = pd.read_csv(MODEL_RESULTS_DIR / "permutation_feature_importance.csv")

print("Loaded test comparison rows:", len(test_comparison))
print("Loaded feature importance rows:", len(feature_importance))
```

## 2. Final Model Comparison

This table compares the final 2024 test performance. Lower RMSE is better.

Important point: Ridge Regression is the locked trained ML model, but the persistence baseline is slightly better on RMSE.


```python
display_columns = ["Model", "MAE", "MSE", "RMSE", "R2", "Directional_Accuracy", "Locked_Selection"]
summary = test_comparison[display_columns].copy()
summary[["MAE", "MSE", "RMSE", "R2", "Directional_Accuracy"]] = summary[["MAE", "MSE", "RMSE", "R2", "Directional_Accuracy"]].round(4)
display(summary)
```

### Speaking note

Ridge Regression test RMSE is 2.8764 and R2 is 0.9873, so it follows the overall price level well. However, the persistence baseline RMSE is 2.8606, which is slightly better. This shows that today's close is already a very strong predictor for tomorrow's close.


## 3. RMSE Comparison Graph

This graph makes the model comparison easier to explain visually.


```python
plot_df = test_comparison.sort_values("RMSE")
plt.figure(figsize=(8, 4.5))
colors = ["tab:orange" if model == "Ridge Regression" else "steelblue" for model in plot_df["Model"]]
plt.barh(plot_df["Model"], plot_df["RMSE"], color=colors)
plt.xlabel("Test RMSE")
plt.title("2024 Test RMSE Comparison (Lower is Better)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 4. Feature Importance

Permutation importance checks how much RMSE increases when one feature is shuffled. If RMSE increases a lot, the feature is important.


```python
top_features = feature_importance.head(8).copy()
top_features[["Importance_Mean", "Importance_Std"]] = top_features[["Importance_Mean", "Importance_Std"]].round(4)
display(top_features)

plt.figure(figsize=(8, 4.5))
plot_features = feature_importance.head(8).sort_values("Importance_Mean")
plt.barh(plot_features["Feature"], plot_features["Importance_Mean"], color="teal", alpha=0.8)
plt.xlabel("Increase in RMSE after permutation")
plt.title("Top Feature Importance - Ridge Regression")
plt.tight_layout()
plt.show()
```

### Speaking note

The most important features are recent price-level variables: Close, Low, Open, MA7, and High. This makes sense because next-day stock price-level prediction is highly related to recent prices.
